**Brute Force solution**

```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]

def find_missing_better(nums):
    result = []
    for i in range(1, len(nums) + 1):
        if i not in nums:
            result.append(i)
    return result


print(find_missing_better(nums))
```

**BETTER**



```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]
def find_missing_better(nums):
    seen = set(nums)
    result = []
    for i in range(1, len(nums) + 1):
        if i not in seen:
            result.append(i)
    return result


print(find_missing_better(nums))
```

```python
nums1 = [1,2,3,4,5,7,8,1]

def find_missing_optimal(nums):
    for i in range(len(nums)):
        idx = abs(nums[i]) - 1          # value -> 0-based index
        if nums[idx] > 0:
            nums[idx] = -nums[idx]      # mark as visited

    result = []
    for i in range(len(nums)):
        if nums[i] > 0:                 # still positive = never visited
            result.append(i + 1)
    return result

print(find_missing_optimal(nums1))
```
